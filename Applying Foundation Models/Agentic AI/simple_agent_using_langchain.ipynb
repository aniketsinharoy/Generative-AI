## Agentic AI

### Load all required modules

```python
from langchain_cohere import ChatCohere
from langchain_community.tools import DuckDuckGoSearchRun
from dotenv import load_dotenv
load_dotenv()
```

### Creating a simple tool for our agent to use it

```python
serach_tool = DuckDuckGoSearchRun()

serach_tool.invoke('Today IPL 2026 match')
```

### Creating a LLM for our agent to use it 

```python
llm = ChatCohere(model='command-a-03-2025')

llm.invoke('What is white?')
```

### Get ReAct agenet prompt from Hub

```python
from langchain_classic import hub
prompt = hub.pull('hwchase17/react')
```

### Create Agent (ReAct Type Agent)

```python
from langchain_classic.agents import create_react_agent, AgentExecutor

agent = create_react_agent(llm=llm, tools=[serach_tool], prompt=prompt)
```

### Create Agent Executor

```python
agent_executor = AgentExecutor(
    agent=agent, 
    tools=[serach_tool], 
    verbose=True            #It will print the thought process of the agent while executing the task
)
```

### Run Agent

```python
response = agent_executor.invoke({'input':'News about IPL 2026?'})
print(response)
```
