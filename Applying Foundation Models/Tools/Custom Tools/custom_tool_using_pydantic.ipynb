### Create Tool With Pydantic
##### Pydantic help in more strict data typing enforcement

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class AddNumbers(BaseModel):
    a : int = Field(required=True, description="First Number to be added")
    b : int = Field(required=True, description="Second Number to be added")
```

```python
def add_numbers(a,b):
    return a+b
```

```python
add_tool = StructuredTool.from_function(
    func=add_numbers,
    name="Add Numbers",
    description="Add 2 numbers",
    args_schema=AddNumbers
)
```

```python
result = add_tool.invoke({'a':5, 'b':10})
print(result)
```

```python
print(add_tool.name)
print(add_tool.description)
print(add_tool.args)
print(add_tool.args_schema.model_json_schema())  #LLM mainly read this schema to understand how to call the tool
```
