### Create Custom Tools using BaseTool 

```python
from langchain_core.tools import BaseTool
from pydantic import BaseModel, Field
from typing import Type
```

```python
class AddNumbersInput(BaseModel):
    a : int = Field(required=True, description="First Number to be added")
    b : int = Field(required=True, description="Second Number to be added")
```

```python
class AddTool(BaseTool):
    name : str = "Add"
    description : str = "Add 2 numbers"
    args_schema : Type[BaseModel] = AddNumbersInput

    def _run(self, a: int, b: int) -> int:
        return a+b
```

```python
add_tool = AddTool()
result = add_tool.invoke({'a':12, 'b':11})
print(result)
```

```python
print(add_tool.name)
print(add_tool.description)
print(add_tool.args)
print(add_tool.args_schema.model_json_schema())  #LLM mainly read this schema to understand how to call the tool
```
