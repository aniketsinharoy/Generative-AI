### Create Tool With Annotation

```python
from langchain_core.tools import tool
```

```python
#Create a function
#Put docstring to explain what the function does
#Use typing annotations to specify the types of the parameters and return value

def add_numbers(a: int, b: int) -> int:
    """Add two numbers together."""
    return a + b
```

```python
#Tool annotation to make the function a tool

@tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers together."""
    return a + b
```

```python
result = add_numbers.invoke({'a': 5, 'b': 10})
print(result)
```

```python
print(add_numbers.name)
print(add_numbers.description)
print(add_numbers.args)
print(add_numbers.args_schema.model_json_schema())  #LLM mainly read this schema to understand how to call the tool
```
