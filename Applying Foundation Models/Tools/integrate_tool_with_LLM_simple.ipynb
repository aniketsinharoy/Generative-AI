```python
from langchain_core.tools import tool
from langchain_cohere import ChatCohere
from langchain_core.messages import HumanMessage, AIMessage, ToolMessage
from dotenv import load_dotenv
load_dotenv()
```

```python
#Created a simple tool

@tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers together."""
    return a + b
```

```python
# Tool Binding
# Binding tools help LLM to understand it has these tools which it can use. 

llm = ChatCohere(model='command-a-03-2025')
llm_with_tool = llm.bind_tools([add_numbers])
```

```python
# To Store Message history
messages = []
```

```python
# User Query
query = "What is the sum of 5 and 7?"

messages.append(HumanMessage(content=query))
```

```python
# Tool Calling
# LLM recognize if its need to use the tool to answer the question or not

response = llm_with_tool.invoke(messages)
messages.append(response)
print(response)

#Since LLM recognize it needs to use the tool, in reponse it will show a tool_calls which will tell which tools the LLM think it needs to use
print(response.tool_calls)
```

```python
# Tool Execution
# We need to execute the tool, LLM can't execute the tool

#Way 1: Just passing the args to the tool and execute it
tool_response = add_numbers.invoke(response.tool_calls[0]['args'])
print(tool_response)

#Way 2: response.tool_calls return a list of ToolMessages, so we will pass the entire ToolMessage
tool_response = add_numbers.invoke(response.tool_calls[0])
messages.append(tool_response)
print(tool_response)
```

```python
response = llm_with_tool.invoke(messages)
print(response.content)
```
