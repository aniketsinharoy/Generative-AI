```python
from langchain_classic.retrievers.contextual_compression import ContextualCompressionRetriever
from langchain_classic.retrievers.document_compressors import LLMChainExtractor
from langchain_cohere import CohereEmbeddings, ChatCohere
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document
from dotenv import load_dotenv
load_dotenv()
```

```python
# Sample Documents
documents = [
    Document(
        page_content="Evergreen plants retain their leaves throughout the year, providing consistent greenery in forests and gardens. Their needle-like or waxy leaves help reduce water loss and allow them to survive harsh winters. These plants continue to grow slowly even in colder months when other vegetation becomes dormant. In the middle of their steady life cycle lies the quiet process of photosynthesis, where chlorophyll in their leaves captures sunlight to convert carbon dioxide and water into glucose and oxygen. This continuous energy production enables evergreens to maintain their foliage year-round. As a result, they play an essential role in stabilizing soil and supporting wildlife habitats.",
        metadata={"title":"Evergreen Plant"}
    ),
    Document(
        page_content="Insect-eating plants, also known as carnivorous plants, thrive in nutrient-poor soils where most plants struggle to survive. They have evolved specialized structures like sticky leaves or pitcher-shaped traps to capture insects for additional nutrients. Despite their unusual diet, they are still green plants that rely on photosynthesis as their primary source of energy. Through photosynthesis, they use sunlight, water, and carbon dioxide to produce sugars necessary for growth. The insects they digest mainly provide nitrogen and other minerals lacking in their environment. This unique adaptation allows them to flourish in bogs, swamps, and other challenging habitats.",
        metadata={"title":"Insect-Eating Plant"}
    )
]
```

```python
#Initialize the Cohere embeddings and FAISS vector store

embedding_model = CohereEmbeddings(model="multilingual-22-12")

vector_store = FAISS.from_documents(
    documents=documents,
    embedding=embedding_model
)
```

```python
# To create a contextual compression retriever, we need to create a base retriever and base compressor first. 

#Base retriever
base_retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 2}
)

#Base Compressor
llm = ChatCohere(model="command-a-03-2025")
compressor = LLMChainExtractor.from_llm(llm=llm)
```

```python
# Creating Contextual Compression Retriever

context_compresion_retriever = ContextualCompressionRetriever(
    base_compressor=compressor,
    base_retriever=base_retriever
)
```

```python
# Query

query = "What is photosynthesis?"
```

```python
# Getting the relevant documents for the query using the contextual compression retriever

relevant_docs = context_compresion_retriever.invoke(query)
```

```python
for doc in relevant_docs:
    print(doc.page_content)
```
